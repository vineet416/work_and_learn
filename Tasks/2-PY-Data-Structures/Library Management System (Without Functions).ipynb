# Library Management System (Without Functions)

## Objective

Create a small library management system using a dictionary to store book information.

> Implement the program without defining custom functions.

## Menu Options

| Option | Action |
|---:|---|
| 1 | Add a book |
| 2 | Borrow a book |
| 3 | Return a book |
| 4 | Search for a book |
| 5 | Display all books |
| 6 | Exit |

## Required Rules

- A book cannot be borrowed more than once at the same time.
- Returning an unavailable or unborrowed book must display an error message.
- Book searches must be case-insensitive.
- The program should validate invalid menu choices and book names.

## Required Output

- Display the total number of available books.
- Display borrowed books separately.
- Show clear success or error messages for each operation.

```python
books = {"math": 1,
         "science": 0,
         "physics": 1,
         "python": 1,
         "database": 0
         }

menu = None

while menu != 6:
    menu = int(input("""
    Select an option between 1-6 where
    1 =  Add a book, 
    2 = Borrow a book, 
    3 = Return a book, 
    4 = Search for a book, 
    5 = Display all books, 
    6 = Exit
    """))

    if menu < 1 or menu > 6:
        print("Please enter a number between 1-6 from given options")
    else:
        if menu == 1:
            add_book = input("Enter book name you want to add: ").strip().lower()
            if add_book in books.keys:
                print("Book is already available")
            else:
                books[add_book] = 1
                print("Book added successfully")
        elif menu == 2:
            borrow = input("Enter book name you want to borrow: ").strip().lower()
            if borrow not in books.keys():
                print("The book is not available")
            else:
                if books[borrow] == 0:
                    print("The book is not available")
                else:
                    books[borrow] = 0
                    print("Book borrow successfull")
        elif menu == 3:
            Return = input("Enter book name you want to return: ").strip().lower()
            if Return not in books.keys():
                print("This is not our book, Please return book borrowed from our library only.")
            else:
                if books[Return] == 1:
                    print("This is not our book, Please return book borrowed from our library only.")
                else:
                    books[Return] = 1
                    print("Book returned successfully")
        elif menu == 4:
            search = input("Enter book name you want to search: ").strip().lower()
            if search not in books.keys():
                print("The book is not available")
            else:
                if books[search] == 0:
                    print("The book is out of stock")
                else:
                    print("The book is available")
        elif menu == 5:
            print(f"All books available have a value 1 and 0 for unavailable \n {books.items()}")
        elif menu == 6:
            break
```
