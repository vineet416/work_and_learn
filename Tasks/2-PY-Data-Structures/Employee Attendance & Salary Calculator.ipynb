# Employee Attendance & Salary Calculator

## Objective

Create a program to calculate employee salaries based on attendance and late days, then display a summary of the company's payroll.

## Employee Details

The company has multiple employees. For each employee, collect the following information:

- **Name**
- **Salary per day**
- **Days present**
- **Late days**

## Salary Calculation

Calculate the gross salary using the employee's daily salary and the number of days present:

$$
\text{Gross Salary} = \text{Salary per Day} \times \text{Days Present}
$$

## Late-Day Deduction Rules

| Late Days | Deduction |
|---|---:|
| 0–2 | No deduction |
| 3–5 | 5% |
| 6–10 | 10% |
| Above 10 | 20% |

Calculate the final salary after applying the deduction:

$$
\text{Final Salary} = \text{Gross Salary} - \text{Deduction}
$$

## Required Output

For each employee, display:

- Gross salary
- Deduction amount
- Final salary

Also display the following company-wide summary:

- Highest salary
- Lowest salary
- Total company payroll
- Employee earning the highest salary

```python
n = int(input("Enter number of employees: "))
employee_details = {}
print("="*50)
print("="*50)
print("\nEmployee Salary Details: \n")
print("="*50)
print("="*50)

if n <= 0:
    print("Number of Employee should be greater then 0")
else:
    for i in range(1, n+1):
        name = input(f"Enter employee {i} name: ")
        salary_per_day = int(input(f"Enter {name}'s salary per day: "))
        days_present = int(input(f"Enter {name}'s number of days present: "))
        late_days = int(input(f"Enter {name}'s number of late days: "))

        # calculate gross salary
        gross_salary = salary_per_day * days_present

        # calculate deduction amount
        if late_days > 10:
            deduction_amount = gross_salary * 0.20
        elif late_days >= 6:
            deduction_amount = gross_salary * 0.10
        elif late_days >= 3:
            deduction_amount = gross_salary * 0.05
        else:
            deduction_amount = 0

        # calculate final salary
        final_salary = gross_salary - deduction_amount

        # Update in employee details
        employee_details[name] = {"name": salary_per_day, 
                                  "days_present": days_present, 
                                  "late_days": late_days,
                                  "gross_salary": gross_salary,
                                  "deduction_amount": deduction_amount,
                                  "final_salary": final_salary}

        # Display salary details for each employee
        print(f"\n{name}'s Salary details")
        print(f"Gross Salary: {gross_salary}")
        print(f"Deduction Amount: {deduction_amount}")
        print(f"Final Salary: {final_salary}\n")
        print("="*50)


    # Calculate company-wide summary
    highest_salary = float('-inf')
    lowest_salary = float('inf')
    total_company_payroll = 0
    employee_highest_eraning = None

    for name, i in employee_details.items():
        total_company_payroll += i["final_salary"]
        if i["final_salary"] > highest_salary:
            highest_salary = i["final_salary"]
            employee_highest_eraning = name
        elif i["final_salary"] < lowest_salary:
            lowest_salary = i["final_salary"]

    # Print company-wide summary
    print("="*50)
    print("\nCompany-Wide Summary: \n")
    print("="*50)
    print("="*50)
    print(f"\nHighest Salary: {highest_salary}")
    print(f"Lowest Salary: {lowest_salary}")
    print(f"Total Company Payroll: {total_company_payroll}")
    print(f"Employee Earning The Highest Salary: {employee_highest_eraning}\n")
    print("="*50)
```
