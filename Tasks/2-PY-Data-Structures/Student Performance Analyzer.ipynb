# Student Performance Analyzer

- Description
    - Create a console-based application that accepts marks of multiple students.

- Requirements:
    - Ask the user for the number of students.
    - For every student:
        - Enter name
        - Enter marks in 5 subjects

- Calculate:
    - Total
    - Percentage
    - Grade (A/B/C/D/F)
        - A >= 85
        - B >= 70 and B < 85
        - C >= 60 and C < 70
        - D >= 40 and D < 60
        - F < 40

- Find:
    - Highest scorer
    - Lowest scorer
    - Average percentage of the class
    - Number of students who failed
    
Display all information in a formatted report.


```python
num_of_stud = int(input("Enter the number of students: "))
stud_marks = {}

# Collect student marks
for i in range(num_of_stud):
    name = input(f"Enter Student Name {i+1}: ")

    marks = input(
        f"Enter marks of student {i+1} in 5 subjects "
        "separated by commas. eg: 68, 75, 90, 83, 39: "
    )

    marks_lst = []

    # Check marks
    for mark in marks.split(","):
        mark = int(mark.strip())

        if mark >= 0 and mark <= 100:
            marks_lst.append(mark)

    # Check that exactly 5 valid marks were entered
    if len(marks_lst) == 5:
        stud_marks[name] = marks_lst
    else:
        print("Please enter exactly 5 valid marks between 0 and 100.")


# Calculate
stud_grades = {}
stud_percentages = []

highest_scorer = None
highest_percentage = -1

lowest_scorer = None
lowest_percentage = 101

no_of_stud_fail = 0

for stud, marks in stud_marks.items():

    # Calculate total and percentage
    total_stud_marks = sum(marks)
    total_marks = 500

    percentage = round((total_stud_marks / total_marks) * 100, 2)

    stud_percentages.append(percentage)

    # Calculate overall grade
    if percentage >= 85:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 40:
        grade = "D"
    else:
        grade = "F"

    stud_grades[stud] = grade

    # Find highest scorer
    if percentage > highest_percentage:
        highest_percentage = percentage
        highest_scorer = stud

    # Find lowest scorer
    if percentage < lowest_percentage:
        lowest_percentage = percentage
        lowest_scorer = stud

    # Find number of students who failed
    if "F" == grade:
        no_of_stud_fail += 1

    # Print student report
    print("=" * 50)
    print(f"\n{stud} Report")
    print(f"Total marks = {total_stud_marks}")
    print(f"Percentage = {percentage}%")
    print(f"Grade = {grade}\n")
    print("=" * 50)


# Calculate average percentage of class
if len(stud_percentages) > 0:
    avg_percentage = round(
        sum(stud_percentages) / len(stud_percentages), 2
    )
else:
    avg_percentage = 0


# Print class report
print("=" * 50)
print(f"\nHighest scorer name: {highest_scorer} -> {highest_percentage}%")
print(f"Lowest scorer name: {lowest_scorer} -> {lowest_percentage}%")
print(f"Average percentage of the class = {avg_percentage}%")
print(
    f"Number of students who failed = "
    f"{no_of_stud_fail}\n"
)
print("=" * 50)
```
