# Student Performance Analyzer

- Description
    - Create a console-based application that accepts marks of multiple students.

- Requirements:
    - Ask the user for the number of students.
    - For every student:
        - Enter name
        - Enter marks in 5 subjects

- Calculate:
    - Total
    - Percentage
    - Grade (A/B/C/D/F)
        - A >= 85
        - B >= 70 and B < 85
        - C >= 60 and C < 70
        - D >= 40 and D < 60
        - F < 40

- Find:
    - Highest scorer
    - Lowest scorer
    - Average percentage of the class
    - Number of students who failed
    
Display all information in a formatted report.


```python
num_of_stud = int(input("Enter the number of students: "))
stud_marks = {}

# Collect student marks
for i in range(num_of_stud):
    name = input(f"Enter Student Name {i+1}: ")
    marks = input(f"Enter marks of student {i+1} in 5 subjects separated by a commas. eg: 68, 75, 90, 83, 39")
    marks_lst = []
    # clean inputs
    for i in marks.split(','):
        mark = int(i.strip())
        if mark < 0 or mark > 100:
            print("All the marks should be between 0-100")
        else:
            marks_lst.append(mark)
    stud_marks[name] = marks_lst
  
# Calculate
stud_grades = {}
highest_scorer = None
highest_percentage = -1
lowest_scorer = None
lowest_percentage = 101
stud_percentages = []
no_of_stud_fail = 0

for stud, mark in stud_marks.items(): 
    total_stud_marks = sum(mark)
    total_marks = len(mark) * 100
    percentage = round((total_stud_marks / total_marks) * 100, 2)
    stud_percentages.append(percentage)

    # calculate grades
    grades = []
    for m in mark:
        if m >= 85:
            grade = "A"
        elif m >= 70:
            grade = "B"
        elif m >= 60:
            grade = "C"
        elif m >= 40:
            grade = "D"
        else:
            grade = "F"
        grades.append(grade)
    stud_grades[stud] = grades

    # Find highest & lowest scorer
    if highest_percentage < percentage:
        highest_percentage = percentage
        highest_scorer = stud
    elif lowest_percentage > percentage:
        lowest_percentage = percentage
        lowest_scorer = stud

    # Find number of students fail
    if "F" in grades:
        no_of_stud_fail += 1

    # Print Report
    print("=" * 50)
    print(f"\n{stud} Report")
    print(f"Total marks = {total_stud_marks}")
    print(f"Percentage =  {percentage}%")
    print(f"Grades = {grades}\n")
    print("=" * 50)


# Calculate Average percentage of class
avg_percentage = round(sum(stud_percentages) / len(stud_percentages), 2)

# Print Report
print("=" * 50)
print(f"\nHighest scorer name: {highest_scorer} -> {highest_percentage}%")
print(f"Lowest scorer name: {lowest_scorer} -> {lowest_percentage}%")
print(f"Average percentage of the class = {avg_percentage}%")
print(f"Number of students who failed in atleast 1 subject = {no_of_stud_fail}\n")
print("=" * 50)
```
