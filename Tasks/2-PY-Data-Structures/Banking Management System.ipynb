# Banking Management System

- Description
    - Build a banking application.

- Menu
    1. Create Account
    2. Deposit
    3. Withdraw
    4. Check Balance
    5. Transfer Money
    6. Show All Accounts
    7. Display total money stored in the bank,
    8. Display richest customer
    9. Exit

- Store account details inside dictionaries.
    - Each account contains
    - Account Number
    - Name
    - Balance

- Rules
    - Cannot withdraw more than balance.
    - Deposit amount must be positive.
    - Transfer should update both accounts.
    - Prevent duplicate account numbers.
    - Show total money stored in the bank.
    - Display richest customer.

```python
# Sample accounts details
account_details = {
    "1001": {
    "Account_Number": "1001",
    "Name": "Vineet Patel",
    "Balance": 12000
},
    "1002": {
    "Account_Number": "1002",
    "Name": "Rohit Sharma",
    "Balance": 45000
},
    "1003": {
    "Account_Number": "1003",
    "Name": "Virat Kolhi",
    "Balance": 18000
    }
}

menu = int(input("""
Enter 1 to Create Account,  
Enter 2 for Deposit,  
Enter 3 for Withdraw,  
Enter 4 to Check Balance,  
Enter 5 to Transfer Money, 
Enter 6 to Show All Accounts,
Enter 7 to Display total money stored in the bank,
Enter 8 to Display richest customer, 
Enter 9 to Exit."""))

if menu < 1 or menu > 9:
    print("Invalid Input. Please enter a number between 1-9")

# Create Account
elif menu == 1:
    account_number = input("Enter your account number")
    if account_number in account_details.keys():
        print("Account number already exist. Please Enter a unique account number.")
    elif account_number not in account_details.keys():
        name = input("Enter your full name.")
        balance = int(input("Enter you available balance."))

        # Add account in account_details
        account_details[account_number] = {"Account_Number": account_number, "Name": name, "Balance": balance}
        print(f"Your account created successfully. Account Number: {account_number}, Name: {name}, Balance: ${balance}")

# Deposit
elif menu == 2:
    account_number = input("Enter the account number in which you want to deposit money.")
    if account_number not in account_details.keys():
        print("Account not found. please enter a valid account number")
    elif account_number in account_details.keys():
        amount = int(input("Enter the amount you want to depost"))
        if amount <= 0:
            print("Please enter amount greater that 0.")
        else:
            # Update Account balance
            account_details[account_number]['Balance'] += amount
            updated_balance = account_details[account_number]['Balance']
            print(f"Deposit sucessfull. Available balance is ${updated_balance}")

# Withdraw
elif menu == 3:
    account_number = input("Enter the account number from which you want to withdraw money.")
    if account_number not in account_details.keys():
        print("Account not found. please enter a valid account number")
    elif account_number in account_details.keys():
        amount = int(input("Enter the amount you want to withdraw"))
        available_Balance = account_details[account_number]["Balance"]
        if amount > available_Balance:
            print(f"Insufficient balance. Available balance to withdraw is ${available_Balance}")
        else:
            # Update Account balance
            account_details[account_number]['Balance'] -= amount
            updated_balance = account_details[account_number]['Balance']
            print(f"Withdraw sucessfull. Available balance is ${updated_balance}")

# Check Balance
elif menu == 4:
    account_number = input("Enter the account number from which you want to check balance.")
    if account_number not in account_details.keys():
        print("Account not found. please enter a valid account number")
    else:
        balance = account_details[account_number]['Balance']
        print(f"Available balance is ${balance}")

# Transfer Money
elif menu == 5:
    sender_account = input("Enter the Sender’s account number from which you want to transfer money.")
    if sender_account not in account_details.keys():
        print("Account not found. please enter a valid account number")
    elif sender_account in account_details.keys():
        receiver_account = input("Enter the receiver’s account number in which you want to transfer money.")
        amount = int(input("Enter the amount you want to transfer"))
        available_Balance = account_details[sender_account]["Balance"]
        if amount > available_Balance:
            print(f"Insufficient balance. Available balance to transfer is ${available_Balance}")
        else:
            # Transfer money
            account_details[sender_account]['Balance'] -= amount
            updated_sender_balance = account_details[sender_account]['Balance']
            account_details[receiver_account]['Balance'] += amount
            updated_receiver_balance = account_details[receiver_account]['Balance']
            print("Transfer successfull.")
            print(f"Sender's account updated balance is ${updated_sender_balance}")
            print(f"Receiver's account updated balance is ${updated_receiver_balance}")

# Show All Accounts
elif menu == 6:
    if not account_details.keys():
        print("No accounts available.")
    else:
        print("Available accounts in our database")
        for i in account_details.keys():
            print(account_details[i])

# Display total money stored in the bank
elif menu == 7:
    if not account_details.keys():
            print("Total money stored in the bank is $0")
    else:
        total_money = 0
        for acc_num in account_details.keys():
            total_money += account_details[acc_num]["Balance"]

        print(f"Total money stored in the bank is ${total_money}")

# Display richest customer
elif menu == 8:
    # Find max balance
    max_balance = 0
    for acc_num in account_details.keys():
        balance = account_details[acc_num]["Balance"]
        if max_balance < balance:
            max_balance = balance

    # find all accounts having max balance
    richest_customers = []
    for acc_num in account_details.keys():
        balance = account_details[acc_num]["Balance"]
        if balance == max_balance:
            richest_customers.append(account_details[acc_num])
    richest_customers

    # Display richest customer details
    for acc in richest_customers:
        print("Name:", acc["Name"])
        print("Account Number:", acc["Account_Number"])
        print("Balance: $", acc["Balance"])

# Exit
elif menu == 9:
    print("Exited successfully.")
```
