# Online Shopping Cart System

## Objective

Create a command-line shopping cart that allows users to add products, calculate totals, apply discounts, and generate an invoice.

## Products and Prices

| Product | Price (₹) |
|---|---:|
| Laptop | 65,000 |
| Phone | 25,000 |
| Mouse | 800 |
| Keyboard | 1,200 |
| Monitor | 9,000 |
| Speaker | 2,500 |

## Required Features

- Allow users to add multiple products to the cart.
- Prevent invalid product selections.
- Ask the user to enter a quantity for each product.
- Calculate the subtotal.
- Apply the appropriate discount based on the subtotal.
- Generate a final invoice.

## Discount Rules

| Subtotal | Discount |
|---|---:|
| Above ₹60,000 | 15% |
| Above ₹30,000 | 10% |
| Above ₹10,000 | 5% |
| ₹10,000 or less | 0% |

> Apply the highest qualifying discount. For example, a subtotal above ₹60,000 receives a 15% discount.

## Invoice Format

| Item | Quantity | Price (₹) | Total (₹) |
|---|---:|---:|---:|
| Product name | Quantity | Unit price | Quantity × unit price |

**Subtotal:** ₹0  
**Discount:** ₹0  
**Final Amount:** ₹0

```python
cart_items = input("""Enter all the products number you want to add to cart seprated by comma 
example "1, 2, 3, 4, 5, 6" where  
1 -> Laptop, 
2 -> Phone, 
3 -> Mouse, 
4 -> Keyboard, 
5 -> Monitor, 
6 -> Speaker
""")

# Clean selected cart items
selected_cart_items = []
for i in cart_items.split(','):
    product = int(i.strip())
    if product < 1 or product > 6:
        print("Invalid input (please enter a numbers between 1-6)")
    else:
        selected_cart_items.append(product)
        

# check if all are valid inputs
# collect product quantity
cart = {}
for i in range(len(selected_cart_items)):
    if i == 1:
        laptop_quantity = int(input("please enter laptop quantity"))
        if laptop_quantity <= 0:
                print("please enter quantity greater than 0")
        else:
            cart["Laptop"] = laptop_quantity
    elif i == 2:
        phone_quantity = int(input("please enter phone quantity"))
        if phone_quantity <= 0:
            print("please enter quantity greater than 0")
        else:
            cart["Phone"] = phone_quantity
    elif i == 3:
        mouse_quantity = int(input("please enter mouse quantity"))
        if mouse_quantity <= 0:
            print("please enter quantity greater than 0")
        else:
            cart["Mouse"] = mouse_quantity
    elif i == 4:
        keyboard_quantity = int(input("please enter keyboard quantity"))
        if keyboard_quantity <= 0:
            print("please enter quantity greater than 0")
        else:
            cart["Keyboard"] = keyboard_quantity
    elif i == 5:
        monitor_quantity = int(input("please enter monitor quantity"))
        if monitor_quantity <= 0:
            print("please enter quantity greater than 0")
        else:
            cart["Monitor"] = monitor_quantity
    elif i == 6:
        speaker_quantity = int(input("please enter speaker quantity"))
        if speaker_quantity <= 0:
            print("please enter quantity greater than 0")
        else: 
            cart["Speaker"] = mouse_quantity


# Calculate subtotal
Subtotal = 0  
print("="*50)
print("\nInvoice\n")
print("="*50)
print("\nProduct    |    Price    |    Quantity\n") 
print("="*50)

for prod, qt in cart.items():
    if prod == "Laptop":
        Subtotal += 65000 * qt
        print(f"\n{prod}    |    ₹60,000    |    {qt}")
    elif prod == "pPhone":
        Subtotal += 25000 * qt
        print(f"\n{prod}    |    ₹25,000    |    {qt}")
    elif prod == "Mouse":
        Subtotal += 800 * qt
        print(f"\n{prod}    |    ₹800    |    {qt}")
    elif prod == "Keyboard":
        Subtotal += 1200 * qt
        print(f"\n{prod}    |    ₹1,200    |    {qt}")
    elif prod == "Monitor":
        Subtotal += 9000 * qt
        print(f"\n{prod}    |    ₹9,000    |    {qt}")
    elif prod == "Speaker":
        Subtotal += 2500 * qt
        print(f"\n{prod}    |    ₹2,500    |    {qt}")


# Calculate Discount
Discount = 0 
if Subtotal > 60000:
    Discount += 0.15* Subtotal
elif Subtotal > 30000:
    Discount += 0.10* Subtotal
elif Subtotal > 10000:
    Discount += 0.05* Subtotal

# Calculate final amount
Final_Amount =  Subtotal - Discount

# Print Invoice
print()
print("="*50)
print(f"\nSubtotal = ₹{Subtotal}")
print(f"Discount = ₹{Discount}")
print(f"Final Amount = ₹{Final_Amount}\n")
print("="*50)
```
