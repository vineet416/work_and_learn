# Online Shopping Cart System

## Objective

Create a command-line shopping cart that allows users to add products, calculate totals, apply discounts, and generate an invoice.

## Products and Prices

| Product | Price (₹) |
|---|---:|
| Laptop | 65,000 |
| Phone | 25,000 |
| Mouse | 800 |
| Keyboard | 1,200 |
| Monitor | 9,000 |
| Speaker | 2,500 |

## Required Features

- Allow users to add multiple products to the cart.
- Prevent invalid product selections.
- Ask the user to enter a quantity for each product.
- Calculate the subtotal.
- Apply the appropriate discount based on the subtotal.
- Generate a final invoice.

## Discount Rules

| Subtotal | Discount |
|---|---:|
| Above ₹60,000 | 15% |
| Above ₹30,000 | 10% |
| Above ₹10,000 | 5% |
| ₹10,000 or less | 0% |

> Apply the highest qualifying discount. For example, a subtotal above ₹60,000 receives a 15% discount.

## Invoice Format

| Item | Quantity | Price (₹) | Total (₹) |
|---|---:|---:|---:|
| Product name | Quantity | Unit price | Quantity × unit price |

**Subtotal:** ₹0  
**Discount:** ₹0  
**Final Amount:** ₹0

```python
products = {
    1: ("Laptop", 65000),
    2: ("Phone", 25000),
    3: ("Mouse", 800),
    4: ("Keyboard", 1200),
    5: ("Monitor", 9000),
    6: ("Speaker", 2500),
}

print("Available products:")
for product_id, (name, price) in products.items():
    print(f"{product_id} -> {name} (₹{price})")

cart_items = input("Enter product numbers separated by commas: ")
selected_cart_items = []

for item in cart_items.split(","):
    item = item.strip()
    try:
        product_id = int(item)
    except ValueError:
        print(f"Invalid product selection: {item}")
        continue

    if product_id not in products:
        print(f"Invalid product number: {product_id}. Please choose 1-6.")
    else:
        selected_cart_items.append(product_id)

cart = {}
for product_id in selected_cart_items:
    product_name, price = products[product_id]

    while True:
        try:
            quantity = int(input(f"Enter quantity for {product_name}: "))
        except ValueError:
            print("Please enter a whole number greater than 0.")
            continue

        if quantity <= 0:
            print("Please enter a quantity greater than 0.")
            continue
        break

    cart[product_id] = cart.get(product_id, 0) + quantity

subtotal = 0
print("\n\n" + "=" * 50)
print("Invoice")
print("=" * 50)
print(f"{'Item'} | {'Quantity'} | {'Price (₹)'} | {'Total (₹)'}")
print("-" * 50)

for product_id, quantity in cart.items():
    product_name, price = products[product_id]
    line_total = price * quantity
    subtotal += line_total
    print(f"{product_name} | {quantity} | {price} | {line_total}")

# Apply the highest qualifying discount.
if subtotal > 60000:
    discount_rate = 0.15
elif subtotal > 30000:
    discount_rate = 0.10
elif subtotal > 10000:
    discount_rate = 0.05
else:
    discount_rate = 0

discount = subtotal * discount_rate
final_amount = subtotal - discount

print("-" * 50)
print(f"Subtotal:     ₹{subtotal}")
print(f"Discount:     ₹{discount}")
print(f"Final Amount: ₹{final_amount}")
print("=" * 50)
```
