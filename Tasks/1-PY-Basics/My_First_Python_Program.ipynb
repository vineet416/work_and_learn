**Write a Python program that demonstrates:**   

Variable declaration    
Different data types    
User input using input()    
Printing output using print()   
Basic arithmetic operations 
String concatenation    
Type conversion (int(), float(), str()) 
Comments in Python  
Multiple print statements explaining each step  

```python
# My First Python Program

# Variable Declaration
name = input("Enter your name: ")
age = int(input("Enter your age: "))
height = float(input("Enter your height in meters: "))

# Printing User Information
print("\nUser Information")
print("Name:", name)
print("Age:", age)
print("Height:", height, "meters")

# Basic Arithmetic Operations
next_age = age + 1
double_age = age * 2
height_cm = height * 100

print("\nArithmetic Operations")
print("Next Year Your Age Will Be:", next_age)
print("Double Your Age:", double_age)
print("Your Height in Centimeters:", height_cm)

# String Concatenation
message = "Hello, " + name + "! Welcome to Python."

print("\nGreeting")
print(message)

# Type Checking
print("\nData Types")
print(type(name))
print(type(age))
print(type(height))

# Type Conversion
age_string = str(age)
height_integer = int(height)

print("\nType Conversion")
print("Age as String:", age_string)
print("Height as Integer:", height_integer)

print("\nThank you for using my first Python program!")
```
