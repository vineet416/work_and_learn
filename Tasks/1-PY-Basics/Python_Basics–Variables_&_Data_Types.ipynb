## Mutable vs Immutable Data Types

Now let's understand an important concept.  

Mutable 
Mutable means the data can be changed after it is created.      

Example:    
List    
Dictionary  
Set 

```python
fruits = ["Apple", "Banana"]
print(fruits)

fruits.append("Mango")
print(fruits)
```

Immutable   
Immutable means the data cannot be changed after it is created. 

Examples:   
Integer 
Float   
String  
Tuple   
Boolean 

```python
colors = ("Red", "Green")

colors[0] = "Blue"
```
