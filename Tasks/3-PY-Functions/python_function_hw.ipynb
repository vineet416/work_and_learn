6. Grade Calculator
- Create a function that accepts marks in five subjects.
- The function must calculate:
	- Total marks
	- Percentage
	- Grade
	- Pass or fail status
- Return all the results from the function.


```python
def total_marks(a:int, b:int, c:int, d:int, e:int)-> int:
    return a+b+c+d+e

def percentage(student_total_marks:int, total_marks:int = 500)-> float:
    return round((student_total_marks / total_marks) *100, 2)

def grade(percentage: float)-> str:
    if percentage >= 85:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 40:
        grade = "D"
    else:
        grade = "F"
    return grade

def pass_or_fail(grade:str) -> str:
    if grade == "F":
        return "Fail"
    else: 
        return "Pass"

def grade_calculator(a:int, b:int, c:int, d:int, e:int)-> str:
    stud_total_marks = total_marks(a,b,c,d,e)
    stud_percentage = percentage(stud_total_marks)
    stud_grade = grade(stud_percentage)
    stud_pass_or_fail = pass_or_fail(stud_grade)
    return f"total marks: {stud_total_marks}, percentage: {stud_percentage}%, grade: {stud_grade}, pass or fail: {stud_pass_or_fail}"

grade_calculator(78,89,62,74,75)
```

- Display both interest amounts and the final payable amounts.

```python
def simple_interest(principal: int, rate:int, time:int) -> int:
    return (principal * rate * time) / 100

def compound_interest(principal: int, rate:int, time:int) -> float:
    return principal * (1+(rate/100)**time)

def calculate_SI_CI(principal: int, rate:int, time:int) -> str:
    SI = simple_interest(principal, rate, time)
    CI = compound_interest(principal, rate, time)
    return f"simple_interest = {SI}, final payable amount for SI = {principal + SI}, compound_interest = {CI}, final payable amount for CI = {principal + CI}"

calculate_SI_CI(100000, 12, 2)
```

12. Prime-Number Checker
- Create a function:
    - is_prime(number)
    - It should return True or False.
- Use this function to check a number entered by the user.


```python
def is_prime(number: int) -> bool:
    if number > 1:
        sqrt = int(number **0.5)
        for i in range(2, sqrt +1):
            if number%i == 0:
                return False
        return True
    else:
        return False

print(is_prime(37))
print(is_prime(49)) 
```

    - is_palindrome(number)
- Use reverse_number() inside is_palindrome().
- Do not convert the number into a string.

```python
def reverse_number(number: int) -> int:
    rev = 0
    while number > 0:
        digit = number % 10  
        rev = rev * 10 + digit
        number //= 10 
    return rev 
    
def is_palindrome(number:int) -> bool:
    if number == reverse_number(number):
        return True
    return False

is_palindrome(1313131)
```

    - Example:
        - 6 = 1 + 2 + 3

```python
def perfect_number_checker(number: int) -> str:
    total = 0
    for i in range(1, 1 + number//2):
        if number % i == 0:
            total += i
    if number == total:
        return f"{number} is a perfect number"
    else:
        return f"{number} is not a perfect number"

perfect_number_checker(28)
```

18. Greatest Common Divisor
- Create a function to find the GCD of two numbers.
- Restriction: Do not use the math.gcd() function.


```python
def gcd(a:int, b:int) -> int:
    divisor = []
    for i in range(1, a + 1):
        if a % i == 0 and b % i == 0:
            divisor.append(i)
    return max(divisor)

gcd(18,12)
```

	- Largest digit

22. Sentence Palindrome
- Create a function to check whether a sentence is a palindrome.
- The function must ignore:
	- Spaces
	- Uppercase and lowercase differences
	- Punctuation marks
- Example:
    - A man, a plan, a canal: Panama


```python
def sentence_palindrome(s:str) -> bool:
    s = s.lower().strip()
    remove_char = "!@#$%^&*()-_=+{}[]|\\:;\"'<>,.?/ "
    result = ""
    
    for i in s:
        if i not in remove_char:
            result += i

    if result == result[::-1]:
        return True
    else: 
        return False

s = "A man, a plan, a canal: Panama"
sentence_palindrome(s)
```

    - Example:
        - Input: [10, 40, 20, 40, 30]
        - Output: 30
- Handle lists that do not contain at least two unique numbers.


```python
def second_highest(l: list) -> int:
    highest = min(l)
    second_highest = min(l)

    for i in l:
        if i > highest:
            highest = i
        elif i > second_highest and i < highest:
            second_highest = i

    if second_highest == min(l):
        return "List does not contain at least two unique numbers"
    else:
        return second_highest


l = [10, 40, 20, 40, 30]
second_highest(l)
```

27. Merge Two Lists
- Create a function that accepts two lists and returns:
	- All unique elements
	- Common elements
	- Elements present only in the first list
	- Elements present only in the second list
- Return the results using a dictionary.


```python
# Without using sets
def merge_two_lists(l1:list, l2:list) -> dict:
    unique_elements = []
    common_elements = []
    elements_only_in_l1 = []
    elements_only_in_l2 = []

    for i in l1:
        if i not in unique_elements:
            unique_elements.append(i)
        if i not in l2:
            elements_only_in_l1.append(i)
        else:
            if i not in common_elements:
                common_elements.append(i)

    for i in l2:
        if i not in unique_elements:
            unique_elements.append(i)
        if i not in l1:
            elements_only_in_l2.append(i)
        else:
            if i not in common_elements:
                common_elements.append(i)

    result = {
        "All unique elements": unique_elements,
	    "Common elements": common_elements,
	    "Elements present only in the first list": elements_only_in_l1,
	    "Elements present only in the second list": elements_only_in_l2
    }
    return result


l1 = [12,24,21,24,21,904,2321]
l2 = [12,34,23,42,64,24,6453]

merge_two_lists(l1, l2)
```

```python
# using sets
def merge_two_lists(l1, l2):
    set1, set2 = set(l1), set(l2)
    return {
        "All unique elements": list(set1 | set2),
        "Common elements": list(set1 & set2), 
        "Elements present only in the first list": list(set1 - set2),
        "Elements present only in the second list": list(set2 - set1)
    }

l1 = [12,24,21,24,21,904,2321]
l2 = [12,34,23,42,64,24,6453]

merge_two_lists(l1, l2)
```

28. Flexible Shopping-Bill Function
- Create a function using *args to accept any number of product prices.
- Use keyword arguments for:
	- Discount percentage
 	- GST percentage
 	- Delivery charge
- The function must return:
 	- Subtotal
 	- Discount
 	- GST
 	- Delivery charge
 	- Final payable amount
- Example call:     
create_bill(    
    500,    
    1200,   
    300,    
    discount=10,    
    gst=18, 
    delivery_charge=50  
)

```python
def create_bill(*args: int, discount:int, gst:int, delivery_charge:int) -> dict:

    product_prices = list(args)
    subtotal = sum(product_prices)
    discount_amount = round((subtotal * discount)/100, 2)
    subtotal_after_discount = subtotal - discount_amount
    gst_amount = round((subtotal_after_discount * gst)/100, 2)
    final_payable_amount = subtotal_after_discount + gst_amount + delivery_charge

    bill = {
        "Product prices": product_prices,
        "Subtotal": subtotal,
        "Discount": discount_amount, 
        "GST": gst_amount,
        "Delivery charge": delivery_charge,
        "Final payable amount": final_payable_amount
    }

    return bill


create_bill(    
    500,    
    1200,   
    300,    
    discount=10,    
    gst=18, 
    delivery_charge=50  
)
```

29. Recursive Folder-Size Simulation

    - Create a recursive function that calculates the total size of all files.

```python
def recursive_folder_size(data: list) -> int:
    total_size = 0
    for i in data:
        if type(i) == int:
            total_size += i
        elif type(i) == list:
            total_size += recursive_folder_size(i)

    return total_size

data = [        
    120,        
    340,        
    [50, 60, [100, 200]],       
    90      
] 

recursive_folder_size(data)
```
