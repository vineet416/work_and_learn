```python
import pandas as pd
```

```python
# Read csv
df = pd.read_csv("products.csv")
```

```python
# Inspect dataset.
df.head()
```

```python
# Find total products.
df.shape[0]
```

```python
# Find unique categories.
df['category'].nunique()
```

```python
# Find unique brands.
df['brand'].nunique()
```

```python
# Count products category-wise.
df.groupby('category')["product_name"].count()
```

```python
# Find average price.
df['price'].mean()
```

```python
# Find highest-priced product.
df[df['price'] == df['price'].max()]
```

```python
# Find lowest-priced product.
df[df['price'] == df['price'].min()]
```

```python
# Find products priced above ₹50,000.
df[df['price'] > 50000]
```

```python
# Find products with rating greater than 4.
df[df['rating'] > 4]
```

```python
# Sort products by price.
df.sort_values(by='price', ascending=False)
```

```python
# Sort products by rating.
df.sort_values(by='rating', ascending=False)
```

```python
# Find products with stock below 10.
df[df['stock'] < 10]
```

```python
# Create discount_amount.
df['discount_amount'] = df['price'] * (df['discount'] / 100)
df
```

```python
# Create final_price.
df['final_price'] = df['price'] - df['discount_amount']
df
```

```python
# Find average price by category.
df.groupby('category')['price'].mean()
```

```python
# Find average rating by brand.
df.groupby('brand')['rating'].mean()
```

```python
# Find maximum product price category-wise.
df.groupby('category')['price'].max()
```

```python
# Export final data.
df.to_csv("output/products_final.csv", index=False)
```
