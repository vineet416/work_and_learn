```python
# Find average Python marks.
df['python_marks'].mean()
```

```python
# Find maximum marks.
df[['python_marks', 'sql_marks', 'pandas_marks']].max()
```

```python
# Sort students by Python marks.
df.sort_values(by='python_marks', ascending=False)
```

```python
# Sort students by attendance.
df.sort_values(by='attendance', ascending=False)
```

```python
# Filter students having Python marks greater than 80.
df[df['python_marks'] > 80]
```

```python
# Filter students having attendance greater than 75.
df[df['attendance'] > 75]
```

```python
# Select only name, python_marks, and pandas_marks.
df[['name', 'python_marks', 'pandas_marks']]
```

```python
# Add a new column called total_marks.
df['total_marks'] = df[['python_marks', 'sql_marks', 'pandas_marks']].sum(axis=1)
df
```

```python
# Add another column called average_marks.
df['average_marks'] = df[['python_marks', 'sql_marks', 'pandas_marks']].mean(axis=1).round(2)
df
```

```python
df.to_csv('output/student_results.csv', index=False)
```
