```python
import pandas as pd
```

```python
# Read the dataset.
df = pd.read_csv('support_tickets.csv')

# Display first 5 rows of the df
df.head()
```

```python
# Find total tickets.
df.shape[0]
```

```python
# Count tickets by category.
total_tickets = df['category'].value_counts()
total_tickets
```

```python
# Count tickets by priority.
total_priority = df['priority'].value_counts()
total_priority
```

```python
# Count tickets by status.
total_status = df['status'].value_counts()
total_status
```

```python
# Find open tickets.
open_tickets = df[df['status'] == 'Open']
open_tickets
```

```python
# Find critical tickets.
critical_tickets = df[df['priority'] == 'Critical']
critical_tickets
```

```python
# Find resolved tickets.
resolved_tickets = df[df['status'] == 'Resolved']
resolved_tickets
```

```python
# Find average resolution time.
average_resolution_time = round(df['resolution_time'].mean(), 2)
average_resolution_time
```

```python
# Find average resolution time per category.
average_resolution_time_by_category = df.groupby('category')['resolution_time'].mean()
average_resolution_time_by_category
```

```python
# Find tickets handled by each support agent.

# Let's first check the unique support agents in the dataset.
unique_agents = df['assigned_agent'].unique()
unique_agents
```

```python
# Find tickets handled by Rahul
rahul_tickets = df[df['assigned_agent'] == 'Rahul']
rahul_tickets
```

```python
# Find tickets handled by Neha
neha_tickets = df[df['assigned_agent'] == 'Neha']
neha_tickets
```

```python
# Find tickets handled by Amit
amit_tickets = df[df['assigned_agent'] == 'Amit']
amit_tickets
```

```python
# Find tickets handled by Priya
priya_tickets = df[df['assigned_agent'] == 'Priya']
priya_tickets
```

```python
# Sort tickets by resolution time.
sorted_tickets = df.sort_values(by='resolution_time', ascending=False)
sorted_tickets
```

```python
# Find tickets taking maximum resolution time.
max_resolution_time = df['resolution_time'].max()
max_resolution_tickets = df[df['resolution_time'] == max_resolution_time]
max_resolution_tickets
```

```python
# Create a summary report of the tickets.
ticket_summary_report = {
    "total_tickets": len(df),
    "average_resolution_time": average_resolution_time,
    "max_resolution_time": max_resolution_time,
    "open_tickets": len(open_tickets),
    "critical_tickets": len(critical_tickets),
    "resolved_tickets": len(resolved_tickets),
    "rahul_tickets": len(rahul_tickets),
    "neha_tickets": len(neha_tickets),
    "amit_tickets": len(amit_tickets),
    "priya_tickets": len(priya_tickets)
}

# Convert the summary report to a DataFrame
ticket_summary = pd.DataFrame(list(ticket_summary_report.items()), columns=['Metric', 'Value'])
ticket_summary
```

```python
# Export the summary report to a CSV file.
ticket_summary.to_csv('ticket_summary.csv', index=False)
```
