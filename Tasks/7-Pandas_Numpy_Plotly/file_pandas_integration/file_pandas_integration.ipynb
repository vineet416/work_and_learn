```python
# Imort the pandas library
import pandas as pd
```

```python
# Read the sales data from the text file into a pandas DataFrame
df = pd.read_csv('sales.txt')

# Display first five rows of the DataFrame
df.head()
```

```python
# type of the DataFrame
type(df)
```

```python
# Shape of the DataFrame
df.shape
```

```python
# DataFrame information
df.info()
```

```python
# Calculate Revenue
df['revenue'] = df['quantity'] * df['price']
df.head()
```

```python
# Find total revenue.
total_revenue = df['revenue'].sum()
total_revenue
```

```python
# Find average order value.
average_order_value = round(df['revenue'].mean(), 2)
average_order_value
```

```python
# Find highest-value order using .loc and .idxmax() methods
highest_value_order = df.loc[df['revenue'].idxmax()]
highest_value_order
```

```python
# Find the order with the highest revenue without using .loc and .idxmax() methods
df[df['revenue'] == df['revenue'].max()]
```

```python
# Find lowest-value order using .loc and .idxmin() methods
lowest_value_order = df.loc[df['revenue'].idxmin()]
lowest_value_order
```

```python
# Find the order with the lowest revenue without using .loc and .idxmin() methods
df[df['revenue'] == df['revenue'].min()]
```

```python
# Find revenue product-wise.
product_revenue = df.groupby('product')['revenue'].sum()
product_revenue
```

```python
# Find revenue city-wise.
city_revenue = df.groupby('city')['revenue'].sum()
city_revenue
```

```python
# Save processed data to a new CSV file
df.to_csv('sales_report.csv', index=False)
```
