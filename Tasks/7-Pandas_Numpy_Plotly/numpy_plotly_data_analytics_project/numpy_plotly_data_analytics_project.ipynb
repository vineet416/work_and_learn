```python
# Import required libraries
import numpy as np
import plotly.express as px
```

```python
# 1. Data Generation
months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])
revenue = np.array([120000, 135000, 142000, 150000, 158000, 165000, 172000, 180000, 195000, 210000, 240000, 280000])
orders = np.array([1200, 1320, 1400, 1480, 1550, 1620, 1700, 1780, 1920, 2050, 2300, 2700])
customers = np.array([850, 920, 1000, 1080, 1160, 1240, 1330, 1420, 1550, 1700, 1900, 2200])
expenses = np.array([75000, 80000, 83000, 87000, 90000, 94000, 97000, 100000, 108000, 115000, 128000, 145000])
```

```python
# 2. NumPy Processing
print("Revenue data type:", type(revenue))
print("Revenue shape:", revenue.shape)
print("Number of months:", revenue.size)
```

```python
# Total annual revenue
total_revenue = np.sum(revenue)
total_revenue
```

```python
# Average monthly revenue
average_monthly_revenue = np.mean(revenue)
round(average_monthly_revenue,2)
```

```python
# Best-performing month
best_month_index = np.argmax(revenue)
best_month = months[best_month_index]
best_revenue = revenue[best_month_index]

print(f"Best-Performing Month: {best_month}")
print(f"Revenue: {best_revenue}")
```

```python
# Lowest-performing month
lowest_month_index = np.argmin(revenue)
lowest_month = months[lowest_month_index]
lowest_revenue = revenue[lowest_month_index]

print(f"Lowest-Performing Month: {lowest_month}")
print(f"Revenue: {lowest_revenue}")
```

```python
# Month-to-month revenue differences
revenue_difference = np.diff(revenue)
revenue_difference
```

```python
# Monthly profit
profit = revenue - expenses
profit
```

```python
# Total annual profit
total_annual_profit = np.sum(profit)
total_annual_profit
```

```python
# 4. PLOTLY VISUALIZATION

# REVENUE TREND
fig = px.line(x=months, y=revenue, markers=True, title="Monthly Revenue Trend", labels={"x": "Month", "y": "Revenue (₹)"})
fig.show()
```

```python
# MONTHLY ORDERS
fig = px.bar(x=months, y=orders, title="Monthly Orders", labels={"x": "Month", "y": "Number of Orders"})
fig.show()
```

```python
# REVENUE VS EXPENSES
plot_months = np.tile(months, 2)
plot_amounts = np.concatenate((revenue, expenses))
plot_categories = np.repeat(["Revenue", "Expenses"], len(months))

fig = px.bar(x=plot_months, y=plot_amounts, color=plot_categories, barmode="group", title="Monthly Revenue vs Expenses", labels={"x": "Month", "y": "Amount (₹)", "color": "Category"})
fig.show()
```

```python
# CUSTOMER GROWTH
fig = px.line(x=months, y=customers, markers=True, title="Monthly Customer Growth", labels={"x": "Month", "y": "Number of Customers"})
fig.show()
```

```python
# MONTHLY PROFIT
fig = px.bar(x=months, y=profit, title="Monthly Profit", labels={"x": "Month", "y": "Profit (₹)"})
fig.show()
```

```python
# MONTH-TO-MONTH REVENUE DIFFERENCE
difference_months = months[1:]

fig = px.bar(x=difference_months, y=revenue_difference, title="Month-to-Month Revenue Difference", labels={"x": "Month", "y": "Revenue Difference (₹)"})
fig.show()
```
