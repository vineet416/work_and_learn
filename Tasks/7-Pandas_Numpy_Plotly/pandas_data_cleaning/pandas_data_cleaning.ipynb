```python
import pandas as pd
```

```python
original_df = pd.read_csv('dirty_customer_data.csv')
```

```python
# Shape of original df
original_df.shape
```

```python
# copy of original_df
df = original_df.copy()
```

```python
# First 5 rows of the df
df.head()
```

```python
# Checking the data types of each column
df.info()
```

```python
# Detecting missing values
df.isnull()
```

```python
# Count of missing values in each column
df.isnull().sum()
```

```python
# Dropping rows with missing values
df.dropna(inplace=True)

# Count of missing values in each column after dropping rows with missing values
df.isnull().sum()
```

```python
# Detecting duplicate rows in the df
df.duplicated()
```

```python
# Count of duplicate rows in the df
df.duplicated().sum()
```

```python
# Deleting duplicate rows in the df
df.drop_duplicates(inplace=True)

# Count of duplicate rows in the df after deleting duplicates
df.duplicated().sum()
```

```python
# Checking the data types of each column
df.dtypes
```

```python
# Correcting data types of columns
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['purchase_amount'] = pd.to_numeric(df['purchase_amount'], errors='coerce')
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
```

```python
# Checking the data types of each column after correcting data types
df.dtypes
```

```python
# Checking for any missing values after correcting data types
df.isnull().sum()
```

```python
# Filling age column missing values with mean
df['age'] = df['age'].fillna(df['age'].mean())

# Filling purchase_amount column missing values with median
df['purchase_amount'] = df['purchase_amount'].fillna(df['purchase_amount'].median())

# Filling rating column missing values with mode
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])
```

```python
# Checking for missing values after filling missing values
df.isnull().sum()
```

```python
# First 5 rows of name column before cleaning
df['name'].head()
```

```python
# Cleaning the name column by removing leading and trailing spaces and converting to title case
df['name'] = df['name'].str.strip().str.title()

# First 5 rows of name column after cleaning
df['name'].head()
```

```python
# First 5 rows of city column before cleaning
df['city'].head()
```

```python
# Cleaning the city column by removing leading and trailing spaces and converting to uppercase
df['city'] = df['city'].str.strip().str.upper()

# First 5 rows of city column after cleaning
df['city'].head()
```

```python
# First 5 rows of email column before cleaning
df['email'].head()
```

```python
# Cleaning the email column by removing leading and trailing spaces and converting to lowercase
df['email'] = df['email'].str.strip().str.lower()

# First 5 rows of email column after cleaning
df['email'].head()
```

```python
# Checking for unique values in city column
df['city'].unique()
```

Above we can see that there is mumbai as well as bombay, which is the same city. We will replace bombay with mumbai.

```python
# Replacing inconsistent city names with correct ones
df['city'] = df['city'].replace({'BOMBAY': 'MUMBAI'})

# Checking for unique values in city column after replacing inconsistent city names
df['city'].unique()
```

```python
# Converting data types of age and purchase_amount columns to integer
df['age'] = df['age'].astype(int)
df['purchase_amount'] = df['purchase_amount'].astype(int)

# Checking data types of all columns after converting data types
df.dtypes
```

```python
# Comparing the shape of original df and cleaned df
print(f"Shape of original df: {original_df.shape}")
print(f"Shape of cleaned df: {df.shape}")
```

```python
# Saving the cleaned df to a new CSV file
df.to_csv('cleaned_customer_data.csv', index=False)
```
