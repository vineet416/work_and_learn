```python
# import required libraries
import plotly.express as px
import numpy as np
```

## 1. Bar Chart

Create a bar chart to represent monthly sales from January to June.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 170, 140, 200, 230]

fig = px.bar(x=months, y=sales, title="Monthly Sales", labels={"x": "Month", "y": "Sales"})
fig.show()
```

## 2. Line Chart

Represent the monthly sales data using a line chart to visualize the sales trend over time.

```python
fig = px.line(x=months, y=sales, markers=True, title="Monthly Sales Trend", labels={"x": "Month", "y": "Sales"})
fig.show()
```

## 3. Pie Chart

Visualize the distribution of employees across different departments.

```python
departments = ["Engineering", "Sales", "Marketing", "HR", "Operations"]
employees = [40, 25, 15, 10, 10]

fig = px.pie(names=departments, values=employees, title="Department-wise Employee Distribution")
fig.show()
```

## 4. Scatter Plot

Analyze the relationship between hours studied and exam marks for 15 students.

```python
students = ["Student 1", "Student 2", "Student 3", "Student 4", "Student 5",
    "Student 6", "Student 7", "Student 8", "Student 9", "Student 10",
    "Student 11", "Student 12", "Student 13", "Student 14", "Student 15"]

hours_studied = [2, 3, 4, 5, 6, 2.5, 3.5, 4.5, 5.5, 6.5, 7, 7.5, 8, 8.5, 9]
exam_marks = [45, 50, 55, 60, 65, 48, 58, 63, 68, 72, 75, 78, 82, 85, 90]

fig = px.scatter(x=hours_studied, y=exam_marks, hover_name=students, 
                 title="Relationship Between Study Hours and Exam Marks",
                 labels={"x": "Hours Studied", "y": "Exam Marks"})
fig.show()
```

## 5. Histogram

Generate 500 random values using NumPy and visualize their distribution using a histogram.

```python
random_values = np.random.normal(loc=50, scale=10, size=500)

fig = px.histogram(x=random_values, nbins=20, title="Distribution of 500 Random Values", labels={"x": "Value","y": "Frequency"})
fig.show()
```

## 7. Multiple Categories

Compare monthly sales of laptops, mobiles, and tablets across six months.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
laptop_sales = [80, 95, 110, 100, 130, 150]
mobile_sales = [150, 170, 160, 180, 200, 220]
tablet_sales = [60, 70, 75, 65, 85, 95]

fig = px.line(x=months, y=laptop_sales, markers=True, title="Monthly Sales Comparison by Product", labels={"x": "Month", "y": "Units Sold"})
fig.add_scatter(x=months, y=mobile_sales, mode="lines+markers", name="Mobile")
fig.add_scatter(x=months, y=tablet_sales, mode="lines+markers",name="Tablet")
fig.update_layout(legend_title="Product")
fig.show()
```

## 8. Student Performance

Compare the marks of 10 students in Python, Mathematics, and Data Science.

```python
students = ["Student 1", "Student 2", "Student 3", "Student 4", "Student 5", "Student 6", "Student 7", "Student 8", "Student 9", "Student 10"]
python_marks = [85, 78, 92, 88, 76, 95, 82, 90, 74, 87]
math_marks = [80, 75, 89, 84, 72, 91, 79, 86, 70, 83]
data_science_marks = [88, 82, 94, 90, 80, 96, 85, 92, 78, 89]

fig = px.bar(x=students, y=python_marks, title="Student Performance Comparison", labels={"x": "Student", "y": "Marks"})
fig.add_bar(x=students, y=math_marks, name="Mathematics")
fig.add_bar(x=students, y=data_science_marks, name="Data Science")
fig.update_layout(barmode="group", legend_title="Subject")
fig.show()
```

## 9. 3D Scatter Plot

Visualize the relationship between age, salary, and experience for 30 employees.

```python
age = np.random.randint(22, 55, 30)
experience = np.random.randint(1, 25, 30)
salary = 30000 + (experience * 4000) + np.random.randint(0, 20000, 30)

fig = px.scatter_3d(x=age, y=experience, z=salary, title="Age, Experience, and Salary Analysis",
    labels={"x": "Age", "y": "Experience (Years)", "z": "Salary"})
fig.show()
```

## 10. Custom Graph

Analyze a fictional e-commerce customer conversion funnel.

```python
stages = ["Website Visitors", "Product Views", "Add to Cart", "Checkout", "Purchases"]
customers = [10000, 6500, 3000, 1800, 1200]

fig = px.funnel(y=stages, x=customers, title="E-commerce Customer Conversion Funnel",
    labels={"x": "Number of Customers", "y": "Funnel Stage"})
fig.show()
```

## 11. NumPy + Plotly

Generate 100 random numbers using NumPy and visualize them using Plotly.

```python
random_numbers = np.random.randint(1, 101, 100)

fig = px.line(x=list(range(1, 101)), y=random_numbers, title="100 Random Numbers",
    labels={"x": "Data Point", "y": "Random Value"}, markers=True)
fig.show()
```

## 12. Dashboard Thinking Challenge

Create four separate visualizations representing the fictional company's revenue, expenses, employees, and customers.

```python
# Employees
departments = ["Engineering", "Sales", "Marketing", "HR", "Operations"]
employee_count = [45, 25, 15, 8, 7]

fig = px.pie(names=departments, values=employee_count, title="Company Employees by Department")
fig.show()
```

```python
# Customers
customers = [1200, 1450, 1700, 2100, 2500, 3000]

fig = px.line(x=months, y=customers, markers=True, title="Monthly Customer Growth", labels={"x": "Month", "y": "Number of Customers"})
fig.show()
```
