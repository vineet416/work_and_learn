```python
# Read Complete Records
with open("students.txt", "r") as file:
    complete_records = file.read()

# Print the complete records
print(complete_records)
```

```python
# Read Individual Lines
with open("students.txt", "r") as file:
    first_line = file.readline()
    second_line = file.readline()

# Print the individual lines
print("First Line:", first_line.strip())
print("Second Line:", second_line.strip())
```

```python
# Read All Lines
with open("students.txt", "r") as file:
    all_lines = file.readlines()

# Print all lines
for line in all_lines:
    print(line.strip())
```

```python
# Count Number of Student Records
with open("students.txt", "r") as file:
    student_count = sum(1 for line in file)

# Print the number of student records
print("Number of Student Records:", student_count)
```

```python
# Copy Data Into Backup File
with open("students.txt", "r") as source_file:
    with open("students_backup.txt", "w") as backup_file:
        backup_file.write(source_file.read())
```

```python
# Create File Containing Only Names and Courses
name_course_records = []

for line in all_lines:
    parts = line.strip().split(",")
    name = parts[1]
    course = parts[4]
    name_course_records.append(f"{name},{course}\n")
```

```python
with open("names_courses.txt","w") as file:
    file.write("name,course\n")
    file.writelines(name_course_records)
```
