```python
# Create application.log
with open('application.log', 'w') as f:
    f.write(logs)
```

```python
# Read log entries from application.log
with open("application.log", "r") as file:
    log_entries = file.readlines()
```

```python
# Display the Entire Log
for entry in log_entries:
    print(entry.strip())
```

```python
# Count Total Log Entries
total_entries = len(log_entries)
print(f"Total Log Entries: {total_entries}")
```

```python
# Count ERROR Entries
total_error_entries = sum(1 for entry in log_entries if "ERROR" in entry)
print(f"Total ERROR Entries: {total_error_entries}")
```

```python
# Count WARNING Entries
total_warning_entries = sum(1 for entry in log_entries if "WARNING" in entry)
print(f"Total WARNING Entries: {total_warning_entries}")
```

```python
# Count INFO Entries
total_info_entries = sum(1 for entry in log_entries if "INFO" in entry)
print(f"Total INFO Entries: {total_info_entries}")
```

```python
# Create errors.txt
with open('errors.txt', 'w') as f:
    for entry in log_entries:
        if "ERROR" in entry:
            f.write(entry)
```

```python
# Create warnings.txt
with open('warnings.txt', 'w') as f:
    for entry in log_entries:
        if "WARNING" in entry:
            f.write(entry)
```
