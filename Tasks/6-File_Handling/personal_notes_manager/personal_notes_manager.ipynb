```python
# Writing 10 Personal Notes
notes = [
    "Learn Python file handling\n",
    "Practice Python every day\n",
    "Complete my programming assignments\n",
    "Learn Pandas basics\n",
    "Build projects for my portfolio\n",
    "Practice Git and GitHub\n",
    "Learn exception handling\n",
    "Improve problem solving skills\n",
    "Read Python documentation\n",
    "Work on my next project\n"
]
```

```python
# Create and write notes to notes.txt file
file  = open("notes.txt", "w")
file.writelines(notes)
file.close()
```

```python
# Read and display the notes from notes.txt file
file = open("notes.txt", "r")
content = file.read()
file.close()

# Display the content of notes.txt
print(content)
```

```python
# Read a Specific Number of Characters
file = open("notes.txt", "r")

# Read the first 50 characters from the file
first_50_chars = file.read(50)
file.close()

# Display the first 50 characters
print(first_50_chars)
```

```python
# Demonstrate readline()
file = open("notes.txt", "r")

# Read the first line from the file
first_line = file.readline()
file.close()

# Display the first line
print(first_line)
```

```python
# Demonstrate readlines()
file = open("notes.txt", "r")

# Read all lines from the file into a list
all_lines = file.readlines()
file.close()

# Display all lines
for line in all_lines:
    print(line.strip())
```

```python
# Appending additional notes.
additional_notes = [
    "Learn SQL fundamentals\n",
    "Practice data analysis\n",
    "Build an AI project\n"
]

# Append additional notes to notes.txt file
file = open("notes.txt", "a")
file.writelines(additional_notes)
file.close()
```

```python
# Reopen and Display Updated Contents
file = open("notes.txt", "r")
updated_content = file.read()
file.close()

# Display the updated content of notes.txt
print(updated_content)
```

```python
# Demonstrate with open()
with open("notes.txt", "r") as file:
    content_with = file.read()

# Display the content read using with open()
print(content_with)
```
