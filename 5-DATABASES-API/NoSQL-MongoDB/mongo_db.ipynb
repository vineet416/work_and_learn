```python
from pymongo import MongoClient
from pymongo.server_api import ServerApi
import os
from dotenv import load_dotenv

load_dotenv()
# Get the MongoDB connection URI from environment variables
uri = os.getenv("MONGO_DB_URL")

# Create a new client and connect to the server
client = MongoClient(uri, server_api=ServerApi('1'))

# Send a ping to confirm a successful connection
try:
    client.admin.command('ping')
    print("Pinged your deployment. You successfully connected to MongoDB!")
except Exception as e:
    print(e)
```

```python
collection.update_one(
    {"name": "Amit"}, 
    {"$set": {"age": 26}})


stud = collection.find({"name": "Amit"})
print(list(stud))
```

```python
collection.update_many(
    {"age": {"$lt": 25}}, 
    {"$set": {"age": 25}})


print(list(collection.find({"age": {"$lt": 25}})))
print(list(collection.find({"age": {"$eq": 25}})))
```

```python
collection.delete_one({"name": "Amit"})

print(list(collection.find({"name": "Amit"})))
```

```python
collection.delete_many({"age": {"$eq": 25}})

print(list(collection.find({"age": {"$eq": 25}})))
```
