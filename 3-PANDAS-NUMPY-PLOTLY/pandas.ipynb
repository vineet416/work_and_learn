```python
students = pd.DataFrame({ "student_id": [1, 2, 3, 4], "name": ["Aman", "Rahul", "Priya", "Neha"] }) 

marks = pd.DataFrame({ "student_id": [1, 2, 3, 5], "marks": [85, 90, 78, 88] })
```

```python
students
```

```python
marks
```

```python
pd.merge(students, marks, on="student_id", how="inner")
```

```python
pd.merge(students, marks, on="student_id", how="left")
```

```python
pd.merge(students, marks, on="student_id", how="right")
```

```python
pd.merge(students, marks, on="student_id", how="outer")
```

```python
pd.merge(students, marks, on="student_id", how="left", indicator=True)
```

```python
result = pd.merge(students, marks, on="student_id", how="left", indicator=True)
result[result["_merge"] == "left_only"] 
```
