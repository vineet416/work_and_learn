```python
import plotly.express as px
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 170, 160, 210, 250]
})
```

```python
df
```

```python
px.line(df, x='Month', y='Sales', title='Monthly Sales', markers=True)
```

```python
df = pd.DataFrame({
    "Course": ["Python", "Data Science", "GenAI", "DevOps"],
    "Students": [300, 220, 420, 180]
})
```

```python
px.bar(df, x="Course", y="Students", title="Student Enrollment")
```

```python
# Horizontal Bar Chart
px.bar(df, x="Students", y="Course", title="Student Enrollment", orientation='h')
```
